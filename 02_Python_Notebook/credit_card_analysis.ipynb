```python
# check outliers in numerical columns in cc_details_df
numerical_cols = [
    "annual_fees",
    "customer_acq_cost",
    "credit_limit",
    "total_revolving_bal",
    "total_trans_amt",
    "total_trans_ct",
    "avg_utilization_ratio",
    "interest_earned"
]
```

```python
for col in numerical_cols:
    
    Q1 = cc_details_df[col].quantile(0.25)
    Q3 = cc_details_df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = cc_details_df[
        (cc_details_df[col] < lower_bound) |
        (cc_details_df[col] > upper_bound)
    ]
    
    print(f"{col}: {len(outliers)} outliers")
```

```python
# using boxplot
for col in numerical_cols:
    plt.figure(figsize=(10, 2))
    plt.boxplot(cc_details_df[col], vert=False)
    plt.title(col)
    plt.show()
```

```python
# check outlier values
col = "credit_limit"

Q1 = cc_details_df[col].quantile(0.25)
Q3 = cc_details_df[col].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers = cc_details_df[
    (cc_details_df[col] < lower_bound) |
    (cc_details_df[col] > upper_bound)
]

print(outliers[[col]].sort_values(by=col))
```

```python
# Outliers were detected in customer_acq_cost, credit_limit, total_trans_amt, total_trans_ct, and interest_earned using the IQR method. 
# After reviewing the values, they were determined to be valid business observations rather than errors.Therefore, no outliers were removed.
```

```python
# Customer Analysis
# Which age group generates the highest transaction amount?
merged_df["age_group"] = pd.cut(
    merged_df["customer_age"],
    bins=[0, 25, 35, 45, 55, 65, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "65+"]
)
```

```python
merged_df.groupby("age_group", observed=False).agg(
    customers=("client_num", "count"),
    transaction_amount=("total_trans_amt", "sum"),
    avg_credit_limit=("credit_limit", "mean"),
    avg_income=("income", "mean")
).sort_values("transaction_amount", ascending=False)
```

```python
# age group 46-55 nd 36-45 is most valuable.
```

```python
# Does gender affect spending behavior?
merged_df.groupby("gender").agg(
    customers=("client_num", "count"),
    total_transaction_amount=("total_trans_amt", "sum"),
    avg_transaction_amount=("total_trans_amt", "mean"),
    avg_credit_limit=("credit_limit", "mean")
)
```

```python
# Male has done more transactio than Female and also transaction amount is higher.
```

```python
# Which jobs generate the most business?
merged_df.groupby("customer_job").agg(
    customers=("client_num", "count"),
    transaction_amount=("total_trans_amt", "sum"),
    avg_income=("income", "mean"),
    interest_earned=("interest_earned", "sum")
).sort_values("transaction_amount", ascending=False)
```

```python
# Businessman generate the most business.
```

```python
# Card & Spending Analysis
# Which card performs best?
merged_df.groupby("card_category").agg(
    customers=("client_num", "count"),
    transaction_amount=("total_trans_amt", "sum"),
    transaction_count=("total_trans_ct", "sum"),
    interest_earned=("interest_earned", "sum"),
    avg_credit_limit=("credit_limit", "mean")
).sort_values("transaction_amount", ascending=False)
```

```python
# Blue card Performs best. Because it earned more interest than any other card, and transaction count is also higher.
```

```python
# Which expense type has the highest spending?
merged_df.groupby("exp_type").agg(
    transaction_amount=("total_trans_amt", "sum"),
    transaction_count=("total_trans_ct", "sum")
).sort_values("transaction_amount", ascending=False)
```

```python
# Bills has the highest spending than any oother expense type. 
```

```python
# Which payment method is most valuable?
merged_df.groupby("use_chip").agg(
    customers=("client_num", "count"),
    transaction_amount=("total_trans_amt", "sum"),
    transaction_count=("total_trans_ct", "sum")
).sort_values("transaction_amount", ascending=False)
```

```python
# Swipe payment method is most valuable.
```

```python
# Revenue & Customer Value Analysis
# Which customers generate the most interest?
merged_df.groupby("card_category").agg(
    total_interest=("interest_earned", "sum"),
    avg_interest=("interest_earned", "mean"),
    total_transaction_amount=("total_trans_amt", "sum")
).sort_values("total_interest", ascending=False) 
```

```python
# Blue card holder customers generates most interest.
```

```python
# Income Group Analysis
merged_df["income_group"] = pd.qcut(
    merged_df["income"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
merged_df.groupby("income_group", observed=False).agg(
    customers=("client_num", "count"),
    avg_credit_limit=("credit_limit", "mean"),
    transaction_amount=("total_trans_amt", "sum"),
    interest_earned=("interest_earned", "sum")
).sort_values("transaction_amount", ascending=False) 
```

```python
# Very High Income Group generates most value.
```

```python
# Risk & Delinquency Analysis
# Overall delinquency rate
merged_df["delinquent_acc"].mean() * 100
```

```python
# Delinquency by age group
merged_df.groupby("age_group", observed=False).agg(
    customers=("client_num", "count"),
    delinquency_rate=("delinquent_acc", "mean")
) 
```

```python
# in percentage
result = merged_df.groupby(
    "age_group",
    observed=False
)["delinquent_acc"].mean() * 100

print(result)
```

```python
# Delinquency by income group 
merged_df.groupby(
    "income_group",
    observed=False
)["delinquent_acc"].mean().mul(100)
```

```python
# Delinquency by utilization
merged_df["utilization_group"] = pd.cut(
    merged_df["avg_utilization_ratio"],
    bins=[0, 0.3, 0.6, 0.8, 1],
    labels=["Low", "Medium", "High", "Very High"],
    include_lowest=True
)
```

```python
merged_df.groupby(
    "utilization_group",
    observed=False
)["delinquent_acc"].mean().mul(100)
```

```python
# Customer Satisfaction Analysis
# Satisfaction by card category
merged_df.groupby("card_category").agg(
    avg_satisfaction=("cust_satisfaction_score", "mean"),
    customers=("client_num", "count"),
    transaction_amount=("total_trans_amt", "sum")
)
```

```python
# Silver card holder are more satisfied than other card holders.
```

```python
# Satisfaction by contact method
merged_df.groupby("contact").agg(
    avg_satisfaction=("cust_satisfaction_score", "mean"),
    customers=("client_num", "count")
)
```

```python
# customer who have telephone contact method are more satisfied.
```

```python
# Satisfaction by customer job
merged_df.groupby("customer_job").agg(
    avg_satisfaction=("cust_satisfaction_score", "mean"),
    customers=("client_num", "count")
).sort_values("avg_satisfaction", ascending=False)
```
